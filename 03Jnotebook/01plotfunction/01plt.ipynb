```python
print("hello duniya from python script")
#plt.plot(x, y color='color-name', linestyle='line_style', linewidth='vlue', marker='marker sybole', lable='lable name')
#plt.xlabel('text')
#pltxlim(start, end)
import matplotlib.pyplot as plt

months= [1,2,3,4]
sales =[700, 1000, 1600, 1400]

plt.plot(months, sales, color = 'blue', linestyle = '--', linewidth = '2', marker = '^', label = '2026 sales data')
plt.xlabel('Months')
plt.ylabel('Sales in USD')
plt.title('Sales Data Over Months')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='grey', linestyle=':', linewidth=1)
plt.xlim(1,4)
plt.ylim(0,2000)
plt.xticks([1,2,3,4], ['Jan', 'Feb', 'Mar', 'Apr'])
plt.show()
```

```python
#plt.bar(x, y, height, color='color-name', width=value, label='label name')
import matplotlib.pyplot as plt

products = ['A', 'B', 'C', 'D', 'E']
sales = [2500, 3000, 1500, 2000, 2800]

plt.bar(products, sales, color='green', width=0.5, label='2026 Sales Data')
plt.xlabel('Products')
plt.ylabel('Sales in USD')
plt.title('Product-wise Sales Data')
plt.legend(loc='upper right', fontsize=8)
plt.grid(color='grey', linestyle='--', linewidth=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
#plt.pie(values, labels=labels_list, colors=colors_list, autopct='%1.1f%%', startangle=value, explode=explode_list)

regions = ['North', 'South', 'East', 'West']
sales = [3000, 2500, 2000, 1500]

plt.pie(sales, labels=regions, autopct='%1.1f%%', colors=['red', 'blue', 'green', 'orange'], )
plt.title('Sales Distribution by Region')
plt.show()
```

```python
#plt.hist(data, bins=number_of_bins, color='color-name', edgecolor='edge-color', alpha=value)
import matplotlib.pyplot as plt

scores = [55, 67, 45, 78, 89, 90, 66, 72, 81, 95, 88, 76, 84, 91, 73, 69, 77, 85, 92, 60]
plt.hist(scores, bins=5, color='purple', edgecolor='black', alpha=0.7)
plt.xlabel('Score Ranges')
plt.ylabel('Number of Students')
plt.title('Distribution of Student Scores')
plt.grid(color='grey', linestyle='--', linewidth=0.5)
plt.show()
```
