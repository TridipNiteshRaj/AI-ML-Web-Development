```python
#plt.subplot(nrows, ncols, index)
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [12, 15, 13, 17, 19]

plt.subplot(1, 2, 1) #1 row , 2 columns, 1first subplot
plt.plot(x, y, color='red')
plt.title('Line chart')

plt.subplot(1, 2, 2) #1 row , 2 columns, 2second subplot
plt.bar(x,y, color='blue')
plt.title('Bar chart')

plt.tight_layout() # Adjusts subplot params so that subplots fit into figure area.
plt.show()
```

```python
import matplotlib.pyplot as plt

#fig, ax = plt.subplots(nrows, ncols, figsize=(width, height))
fig, ax = plt.subplots(1, 2, figsize=(10, 5)) # 1 row, 2 columns

x = [1, 2, 3, 4, 5]
y = [12, 15, 13, 17, 19]

ax[0].plot(x,y, color='blue')
ax[0].set_title('Line chart')

ax[1].bar(x,y, color='orange')
ax[1].set_title('Bar chart')

#fig, ax = plt.subplots(2, 1, sharex=True) # 2 rows, 1 column
fig.suptitle('Comparison of Line and Bar Charts')

plt.tight_layout() # Adjusts subplot params so that subplots fit into figure area.
plt.show()
```
